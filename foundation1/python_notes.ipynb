##1.List与Dict基础容器

```python
# 切片语法 list[start:end:step]，end索引对应的元素不包含在内
my_list = [0, 1, 2, 3, 4, 5, 6]
print(my_list[1:4])   # 获取索引1到3的元素
print(my_list[::-1])  # 反转整个列表

# 常用内置方法
my_list.append(7)        # 在列表末尾添加单个元素
my_list.extend([8, 9])  # 追加一个可迭代对象的全部元素
my_list.insert(2, 99)   # 在指定索引位置插入元素
my_list.pop()            # 删除并返回末尾元素，可传入索引删除指定位置
my_list.remove(99)       # 删除列表中第一个匹配该值的元素
my_list.index(3)         # 查询元素第一次出现的索引
my_list.count(2)         # 统计元素出现的次数
my_list.sort()           # 原地排序，直接修改原列表
sorted(my_list)          # 返回新的排序列表，不会修改原列表

# 列表解包技巧
a, b, *rest = [10, 20, 30, 40, 50]
print(a, b, rest)
```

```python
# 创建字典
student = {"name": "Alice", "age": 18, "major": "CS"}

# .get() 安全取值，当key不存在时不会抛出KeyError，可以设置默认返回值
print(student.get("name"))
print(student.get("score", 0))

# 不同方式遍历字典
for k in student.keys():
    pass
for v in student.values():
    pass
for k, v in student.items():
    print(k, v)

# 更新字典
student.update({"age": 19, "gender": "F"})

# Python3.9+ 使用 | 运算符合并字典
dict_a = {"x": 1}
dict_b = {"y": 2}
dict_c = dict_a | dict_b
```

##2.Lambda/装饰器

```python
# 普通def函数
def add(x, y):
    return x + y

# 等价lambda表达式
add_lambda = lambda x, y: x + y
print(add_lambda(3, 5))

# 搭配高阶函数使用示例
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x ** 2, nums))
print(squares)

people = [("Bob", 20), ("Alice", 18)]
people_sorted = sorted(people, key=lambda item: item[1])
print(people_sorted)
```

##3.面向对象编程OOP

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        print(f"开始执行函数 {func.__name__}")
        result = func(*args, **kwargs)
        end = time.time()
        print(f"执行完成，耗时: {end - start:.4f} s")
        return result
    return wrapper

@timer
def slow_func():
    time.sleep(1)
    print("函数内部执行完毕")

slow_func()
```

```python
class Dog:
    # 类属性：所有实例对象共享
    species = "Canis lupus familiaris"

    def __init__(self, name: str, age: int):
        # 实例属性：每个对象拥有独立的数据
        self.name = name
        self.age = age

    def bark(self):
        print(f"{self.name}: Woof!")

# 创建实例
my_dog = Dog("Lucky", 3)
my_dog.bark()
print(my_dog.name, my_dog.age)
```

```python
class Book:
    def __init__(self, title: str):
        self.title = title

    def __str__(self):
        return f"Book(title={self.title})"

    def __repr__(self):
        return f"<Book object: {self.title}>"

b = Book("Python学习笔记")
print(b)
```

```python
class Animal:
    def eat(self):
        print("Animal is eating")

class Cat(Animal):
    def meow(self):
        print("Meow~")

kitty = Cat()
kitty.eat()
kitty.meow()
```

##4.文本处理：re正则表达式

```python
import re

text = "user123 score: 98 user456 score: 87"

# compile预编译正则模式，多次使用时效率更高
pattern = re.compile(r"\d+")  # \d+ 匹配一串连续数字

# findall 返回全部匹配结果组成的列表
all_nums = pattern.findall(text)
print(all_nums)

# search 查找第一个匹配项，返回match对象，无匹配返回None
match = pattern.search(text)
if match:
    print(match.group())

# sub 实现文本替换
new_text = pattern.sub("NUMBER", text)
print(new_text)
```

##5.列表推导式/typehint

```python
# 生成0~9数字的平方
squares = [i ** 2 for i in range(10)]
print(squares)

# 增加条件筛选，只保留偶数的平方
even_sq = [i ** 2 for i in range(10) if i % 2 == 0]
print(even_sq)

# 拓展：字典推导式、集合推导式
dict_comp = {i: i ** 2 for i in range(5)}
set_comp = {i % 3 for i in range(10)}
```

```python
# 变量注解
name: str = "Tom"
age: int = 20
scores: list[int] = [80, 90, 95]

# 函数注解：参数类型 -> 返回值类型
def calculate_sum(a: int, b: int) -> int:
    return a + b

# 复杂类型示例
from typing import List, Dict, Optional
def get_user(user_id: int) -> Optional[Dict[str, str]]:
    if user_id == 1:
        return {"name": "Alice"}
    return None
```

##6.生成器/关键字

```python
def my_generator(n: int):
    i = 0
    while i < n:
        yield i
        i += 1

gen = my_generator(5)
# for循环遍历生成器
for num in gen:
    print(num)

# 使用next手动获取值
gen2 = my_generator(3)
g = iter(gen2)
print(next(g))
print(next(g))
print(next(g))
```
