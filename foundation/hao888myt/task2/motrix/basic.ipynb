1. 初始化数组

```python
import numpy

data_matrix = numpy.arange(100).reshape((10, 10))
print(data_matrix)
```

2. 取出中间4x4的矩阵

```python
sub_matrix = data_matrix[3:7,3:7]
print(sub_matrix)
```

3. 定位大于75的元素的位置，并将其替换为0

```python
condition = data_matrix > 75

print(numpy.argwhere(condition))

data_matrix[condition] = 0

print(data_matrix)
```

4. 向量化运算

```python
data_matrix = data_matrix.astype(numpy.float64)
data_matrix *= 0.8

print(data_matrix)
```

5. 获取并定位最大值

```python
pos = data_matrix.argmax()
pos = numpy.unravel_index(pos, data_matrix.shape)

print(f"{pos[0]}, {pos[1]}")
print(data_matrix[pos])
```
