# 广播机制

## 1. 数据生成

```python
import numpy as np

points_A = np.random.randint(0, 101, size=(5, 2))
points_B = np.random.randint(0, 101, size=(8, 2))

print(points_A)
print(points_B)
```

## 2. 距离矩阵计算 be like:



```python
points_A_3d = points_A[:, np.newaxis, :]
points_B_3d = points_B[np.newaxis, :, :]

distance_matrix = np.sqrt(((points_A_3d - points_B_3d) ** 2).sum(axis=2))
print(distance_matrix)
```

## 3. 轴向数据聚合

```python
min_distance = distance_matrix.min(axis = 1)
print(min_distance)
```

## 4. 复合条件查询

```python
condition = (distance_matrix < 20).any(axis = 0)

index = np.flatnonzero(condition)

print(index)
print(points_B[index])
```
