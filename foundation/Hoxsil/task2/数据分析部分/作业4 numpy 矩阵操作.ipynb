# 0. 导入库

```python
import numpy as np
```

# 1. 数据生成

```python
points_A = np.random.randint(0, 100, (5, 2))
points_B = np.random.randint(0, 100, (8, 2))
print("points_A:\n", points_A)
print("points_B: \n", points_B)
```

# 2. 距离矩阵计算

```python
points_A = points_A.reshape(5, 1, 2)
points_B = points_B.reshape(1, 8, 2)
# 写法一：
# distance_matrix = np.sqrt((points_A[:,:,[0]] - points_B[:,:,[0]])**2 + (points_A[:,:,[1]] - points_B[:,:,[1]])**2).squeeze()
# 写法二：
diff = points_A - points_B
distance_matrix = np.sqrt(np.sum(diff**2, axis=-1))
# print("points_A:\n", points_A)
# print("points_B: \n", points_B)
print(distance_matrix)
print(distance_matrix.shape)
```

# 3. 轴向数据聚合

```python
min_A_distance_matrix = np.min(distance_matrix, axis=1)
print(min_A_distance_matrix)
```

# 4. 复合条件查询

```python
# 方法一：
min_B_distance_matrix = np.min(distance_matrix, axis=0)
print(min_B_distance_matrix)
mask = min_B_distance_matrix < 20
index = np.where(mask)
print(index)
```

```python
# 方法二：
mask = distance_matrix < 20
mask = np.any(mask, axis=0)
index = np.where(mask)
print(index)
```
