# 0. 导入库

```python
import numpy as np
```

# 1. 数组创建

```python
data_matrix = np.arange(0, 100).reshape(10, 10)
print(data_matrix)
```

# 2. 子矩阵提取

```python
part_of_data_matrix = data_matrix[3:7, 3:7]
print(part_of_data_matrix)
```

# 3. 条件筛选与赋值

```python
mask = data_matrix > 75
data_matrix[mask] = 0
print(data_matrix)
```

# 4. 向量化运算

```python
data_matrix = data_matrix * 0.8
print(data_matrix)
```

# 5. 找到最终矩阵中的最大值

```python
coordinate = np.unravel_index(np.argmax(data_matrix), data_matrix.shape)
print(coordinate[0], coordinate[1])
print(data_matrix[coordinate])
```
