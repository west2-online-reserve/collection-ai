```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def logistic(x, L, k, x0):
    return L / (1 + np.exp(-k * (x - x0)))

stop = 10
gap = 0.1
x = np.arange(0, stop, gap)

y1 = logistic(x, 100, 1, 50)
y2 = logistic(x, 70, 1, 50)
y3 = logistic(x, 100, 2, 50)
y4 = logistic(x, 100, 1, 70)

figures, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, y1)
axes[0, 0].set_title('y1')
axes[0, 1].plot(x, y2)
axes[0, 1].set_title('y2')
axes[1, 0].plot(x, y3)
axes[1, 0].set_title('y3')
axes[1, 1].plot(x, y4)
axes[1, 1].set_title('y4')
figures.tight_layout()

plt.show()
```
