```python
import numpy as np
```

数据生成

```python
point_A = np.random.randint(0,101,(5,2))
point_B = np.random.randint(0,101,(8,2))
print(point_A,point_B)
```

距离矩阵计算：

```python
#原来newaxis可以这么用(没思路，参考了chatgpt)
diff = point_A[:, np.newaxis, :] - point_B[np.newaxis, :, :]
distance_matrix = np.sqrt(np.sum(diff ** 2, axis=-1))
print(distance_matrix)
```

轴向数据聚合：

```python
min_distance = np.min(distance_matrix,axis=1).reshape(-1)
print(min_distance)
```

```python
print(np.any(distance_matrix<20,axis=0))
min20 =point_B[np.any(distance_matrix<20,axis=0)]
print(min20)
idx_array = np.where(np.any(distance_matrix<20,axis=0))
print(idx_array)
```
