```python
from matplotlib import pyplot as plt
import numpy as np
import math
```

Logistic functton 

```python
L = 1
k = 2
x0 =0
def Logistic_func(x:np.array,L:float,k:float,x0:float)->np.array:
    son = L
    mother = 1 + math.e ** (-k*(x - x0))
    return son / mother 
x = np.linspace(-6,6,2000)
L1=1
k1=1
x01=0

L2=1
k2=5
x02=0

L3=5
k3=1
x03=0

L4=1
k4=1
x04=1

y1 = Logistic_func(x,L1,k1,x01)
y2 = Logistic_func(x,L2,k2,x02)
y3 = Logistic_func(x,L3,k3,x03)
y4 = Logistic_func(x,L4,k4,x04)
fig_1 = plt.figure(figsize=(10,9),dpi = 260)
plt.subplot(2,2,1)
plt.xlabel('x')
plt.ylabel('y')
plt.title('L=1,K=1,x0=0')
fig_1 = plt.plot(x,y1)
plt.subplot(2,2,2)
plt.xlabel('x')
plt.ylabel('y')
plt.title('L=1,K=5,x0=0')
fig_1 = plt.plot(x,y2)
plt.subplot(2,2,3)
plt.xlabel('x')
plt.ylabel('y')
plt.title('L=5,K=1,x0=0')
fig_1 = plt.plot(x,y3)
plt.subplot(2,2,4)
plt.xlabel('x')
plt.ylabel('y')
plt.title('L=1,K=1,x0=1')
fig_1 = plt.plot(x,y4)

plt.show()
```
